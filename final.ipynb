```python
model=Sequential([
    Conv2D(32,(3,3),input_shape=(imgwidth,imgheight,3),activation='relu'),
    MaxPooling2D(2),
    Conv2D(64,(3,3),activation='relu'),
    MaxPooling2D(2),
    Conv2D(128,(3,3),activation='relu'),
    MaxPooling2D(2),
    Conv2D(100,(3,3),activation='relu'),
    MaxPooling2D(2),
    Flatten(),
    Dense(200,activation='relu'),
    Dense(128,activation='relu'),
    Dense(64,activation='relu'),
    Dense(output,activation='softmax')
    
           
])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
gen=ImageDataGenerator(validation_split=0.1)
train_data=gen.flow_from_directory(
    dataPath,
    target_size=(imgwidth,imgheight),
    class_mode='categorical',
    subset=('training'),
    batch_size=64,
)
valid_data=gen.flow_from_directory(
    dataPath,
    target_size=(imgwidth,imgheight),
    class_mode='categorical',
    subset=('validation'),
    batch_size=64,
)
```
